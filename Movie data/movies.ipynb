```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("movies.csv",encoding = 'utf-8',engine='python')
data['Genre'] = data['Genre'].str.split(',')
new_data = data.explode('Genre')
info_pop = new_data[["Title","Popularity"]].drop_duplicates()
top_10 = info_pop.sort_values(by = "Popularity", ascending = False).head(10)
plt.figure(figsize = (18,5))
sns.barplot(
    data = top_10,
    x = "Title",
    y = "Popularity"
)
plt.tight_layout
plt.title("Title vs Popularity")
plt.show()

```

```python
count = new_data["Genre"].str.strip().value_counts()
plt.figure(figsize = (6,8))
plt.pie(
    count,
    labels = count.index,
    autopct = "%1.1f%%"
)
plt.tight_layout
plt.show()
```
